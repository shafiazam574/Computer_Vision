```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
print(cv2.__version__)
```

# Task 01: Erosion and Dialation

```python
image = cv2.imread("D:\\VS Code\\Computer_Vision\\fingerprint-sunglasses.jpg")
image_lena = cv2.imread("D:\\VS Code\\Computer_Vision\\lena.jpg")
image_alphabet = cv2.imread("D:\\VS Code\\Computer_Vision\\alphabet_pic.jpg")

image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
image_lena_gray = cv2.cvtColor(image_lena, cv2.COLOR_BGR2GRAY)
image_alphabet_gray = cv2.cvtColor(image_alphabet, cv2.COLOR_BGR2GRAY)

binary_image = cv2.threshold(image_gray, 127, 255, cv2.THRESH_BINARY)[1]
binary_image_lena = cv2.threshold(image_lena_gray, 127, 255, cv2.THRESH_BINARY)[1]


kernal = np.ones((3, 3), np.uint8)

inverted_binary_image = cv2.bitwise_not(binary_image)
inverted_binary_image_lena = cv2.bitwise_not(binary_image_lena)

eroded_image = cv2.erode(inverted_binary_image, kernal, iterations=1)
eroded_image_lena = cv2.erode(inverted_binary_image_lena, kernal, iterations=1)

dilated_image = cv2.dilate(eroded_image, kernal, iterations=1)
dilated_image_lena = cv2.dilate(eroded_image_lena, kernal, iterations=1)

plt.imshow(image_gray, cmap='gray'), plt.title("Gray Image"), plt.axis('off')
plt.show()
plt.imshow(binary_image, cmap='gray'), plt.title("Binary Image"), plt.axis('off')
plt.show()
plt.imshow(inverted_binary_image, cmap='gray'), plt.title("Inverted Binary Image"), plt.axis('off')
plt.show()
plt.imshow(eroded_image, cmap='gray'), plt.title("Eroded Image"), plt.axis('off')
plt.show()
plt.imshow(dilated_image, cmap='gray'), plt.title("Dilated Image"), plt.axis('off')
plt.show()

plt.imshow(image_lena_gray, cmap='gray'), plt.title("Gray Image"), plt.axis('off')
plt.show()
plt.imshow(eroded_image_lena, cmap='gray'), plt.title("Eroded Image"), plt.axis('off')
plt.show()
plt.imshow(dilated_image_lena, cmap='gray'), plt.title("Dilated Image"), plt.axis('off')
plt.show()
```

# Task 02: Opening and Closing

```python
image = cv2.imread("D:\\VS Code\\Computer_Vision\\alphabet_pic.jpg")
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
binary_image = cv2.threshold(image_gray, 127, 255, cv2.THRESH_BINARY)[1]
kernal = np.ones((3, 3), np.uint8)
inverted_binary_image = cv2.bitwise_not(binary_image)
opened_image = cv2.morphologyEx(inverted_binary_image, cv2.MORPH_OPEN, kernal)
closed_image = cv2.morphologyEx(inverted_binary_image, cv2.MORPH_CLOSE, kernal)

plt.imshow(image_gray, cmap='gray'), plt.title("Gray Image"), plt.axis('off')
plt.show()  
plt.imshow(opened_image, cmap='gray'), plt.title("Opened Image"), plt.axis('off')
plt.show()
plt.imshow(closed_image, cmap='gray'), plt.title("Closed Image"), plt.axis('off')
plt.show()
```

# Task 03: Morphological Gradient

```python
image = cv2.imread("D:\\VS Code\\Computer_Vision\\alphabet_pic.jpg")
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
binary_image = cv2.threshold(image_gray, 127, 255, cv2.THRESH_BINARY)[1]
kernal = np.ones((3, 3), np.uint8)
inverted_binary_image = cv2.bitwise_not(binary_image)

gradient_image = cv2.morphologyEx(inverted_binary_image, cv2.MORPH_GRADIENT, kernal)

plt.imshow(image_gray, cmap='gray'), plt.title("Gray Image"), plt.axis('off')
plt.show()
plt.imshow(gradient_image, cmap='gray'), plt.title("Gradient Image"), plt.axis('off')
plt.show()

image2 = cv2.imread("D:\\VS Code\\Computer_Vision\\noise_car.png")
image2_gray = cv2.cvtColor(image2, cv2.COLOR_BGR2GRAY)
binary_image2 = cv2.threshold(image2_gray, 127, 255, cv2.THRESH_BINARY)[1]
kernal = np.ones((3, 3), np.uint8)
inverted_binary_image2 = cv2.bitwise_not(binary_image2)
gradient_image2 = cv2.morphologyEx(inverted_binary_image2, cv2.MORPH_GRADIENT, kernal)

plt.imshow(image2_gray, cmap='gray'), plt.title("Gray Image"), plt.axis('off')
plt.show()  
plt.imshow(gradient_image2, cmap='gray'), plt.title("Gradient Image"), plt.axis('off')
plt.show()
```

# Task 04: Top Hat
