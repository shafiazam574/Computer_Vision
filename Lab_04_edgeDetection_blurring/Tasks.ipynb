# Task 01: Advance Edge Detection Techniques
1. Sobel edge detection
2. Scharr edge detection
3. Laplacian edge detection

```python
image = cv2.imread("D:\\VS Code\\Computer_Vision\\cameraman.png")

if image is None:
    print("Error: Image not found.")
    exit()
    # exit(1) 1 means there is an error, 0 means there is no error and it is empty like exit() then it will be 0
    # exit function is used to exit the program when there is an error or when we want to stop the program.

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

#-----------------------------------------------------------------------------------------------------------------------------
# SOLBEL EDGE DETECTION

sobel_x = cv2.Sobel(gray_image, cv2.CV_64F, 1,0, ksize = 3) 
sobel_y = cv2.Sobel(gray_image, cv2.CV_64F, 0, 1, ksize = 3)
# cv2.CV_64F is the depth of the output image, it is used to avoid overflow and underflow of pixel values, it can be changed to cv2.CV_8U but it will cause overflow and underflow of pixel values.

sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)
# cv2.convertScaleAbs() is used to convert the pixel values of the image to absolute values and then convert them to 8-bit unsigned integers, it is used to avoid overflow and underflow of pixel values.
# overflow and underflow of pixel values means that when we perform some operations on the image, the pixel values may exceed the maximum value of 255 or go below the minimum value of 0, which can cause distortion in the image.

sobel_combined = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)
# cv2.addWeighted() is used to add two images with weights, it takes 5 parameters, the first parameter is the first image, the second parameter is the weight of the first image, the third parameter is the second image, the fourth parameter is the weight of the second image, and the fifth parameter is the scalar added to each sum.
# weight of the parameter means how much we want to contribute to the final image, if we want to contribute more to the first image then we can increase the weight of the first image and decrease the weight of the second image and vice versa.
# scalar added to each sum is used to adjust the brightness of the final image, if we want to increase the brightness of the final image then we can increase the scalar added to each sum and vice versa.

#-----------------------------------------------------------------------------------------------------------------------------
# SCHARR EDGE DETECTION

scharr_x = cv2.Scharr(gray_image, cv2.CV_64F, 1, 0)
scharr_y = cv2.Scharr(gray_image, cv2.CV_64F, 0, 1)

scharr_x = cv2.convertScaleAbs(scharr_x)
scharr_y = cv2.convertScaleAbs(scharr_y)

scharr_combined = cv2.addWeighted(scharr_x, 0.5, scharr_y, 0.5, 0)

#-----------------------------------------------------------------------------------------------------------------------------
# LAPLACIAN EDGE DETECTION

laplacian = cv2.Laplacian(gray_image, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

#-----------------------------------------------------------------------------------------------------------------------------

# CANNY EDGE DETECTION

canny_edges = cv2.Canny(gray_image, 100, 200)

#-----------------------------------------------------------------------------------------------------------------------------
# DISPLAYING THE IMAGES

cv2.imshow("Original Image", image)
cv2.imshow("Sobel", sobel_combined)
cv2.imshow("Scharr", scharr_combined)
cv2.imshow("Laplacian", laplacian)
cv2.imshow("Canny Edges", canny_edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Task 02: Custom Blurring Filters

### np.float32
1. np.float32 is a NumPy data type that represents a 32-bit floating point number. It is used to store decimal values and is commonly used in OpenCV for kernels and image processing operations.

```python
image = cv2.imread("D:\\VS Code\\Computer_Vision\\cameraman.png")

if image is None:
    print("Error: Image not found.")
    exit()

kernal_5 = np.ones((5,5), np.float32) / 25
kernal_9 = np.ones((9,9), np.float32) / 81

blurred_5 = cv2.filter2D(image, -1, kernal_5)
blurred_9 = cv2.filter2D(image, -1, kernal_9)
# cv2.filter2D(src, ddepth, kernel) --> and depth = -1 means that the depth of the output image will be the same as the input image, it can be changed to cv2.CV_8U or cv2.CV_64F but it will cause overflow and underflow of pixel values.
# cv2.filter2D() is used to perform convolution on the image, it takes 3 parameters, the first parameter is the input image, the second parameter is the depth of the output image, and the third parameter is the kernel used for convolution, it is used to blur the image by applying the kernel to the pixel values in the image.


#------------------------------------------------------------------------------------------------------------------------------
# other standard blurring techniques
averaging_blur_5 = cv2.blur(image, (5,5))
gaussian_blur_5 = cv2.GaussianBlur(image, (5,5), 0)
median_blur_5 = cv2.medianBlur(image, 5)
# cv2.blur() is used to perform averaging blur, it takes 2 parameters, the first parameter is the input image and the second parameter is the kernel size, it is used to blur the image by averaging the pixel values in the kernel size.
# cv2.GaussianBlur() is used to perform Gaussian blur, it takes 3 parameters, the first parameter is the input image, the second parameter is the kernel size, and the third parameter is the standard deviation in the X direction, it is used to blur the image by applying a Gaussian function to the pixel values in the kernel size.
# cv2.medianBlur() is used to perform median blur, it takes 2 parameters, the first parameter is the input image and the second parameter is the kernel size, it is used to blur the image by taking the median of the pixel values in the kernel size.
#------------------------------------------------------------------------------------------------------------------------------

# DISPLAYING THE IMAGES

cv2.imshow("Original Image", image)
cv2.imshow("Blurred Image with 5x5 Kernel", blurred_5)
cv2.imshow("Blurred Image with 9x9 Kernel", blurred_9)
cv2.imshow("Averaging Blur", averaging_blur_5)
cv2.imshow("Gaussian Blur", gaussian_blur_5)
cv2.imshow("Median Blur", median_blur_5)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
