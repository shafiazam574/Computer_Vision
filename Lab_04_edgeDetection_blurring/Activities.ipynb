# Activity 01:  Edge Detection
Edge Detection using OpenCV

## Apply Canny edge detection
#### A.  it detects edges in the image by looking for areas of rapid intensity change. The first parameter is the input image, the second and third parameters are the lower and upper thresholds for the hysteresis procedure used in edge detection.
#### B.  The Canny edge detection algorithm works by first applying a Gaussian blur to the image to reduce noise, then finding the intensity gradients of the image, and finally applying non-maximum suppression and hysteresis thresholding to identify the edges.
#### C.  steps :
1. Noise Reduction: The image is smoothed using a Gaussian filter to reduce noise and unwanted details.
2. Gradient Calculation: The intensity gradients of the image are calculated using Sobel operators to find the edges. This step produces a gradient magnitude and direction for each pixel.
3. Non-Maximum Suppression: This step thins the edges by suppressing non-maximum pixels in the gradient direction, resulting in thin edges.
4. Hysteresis Thresholding: Finally, the algorithm uses two thresholds to determine which edges are strong, weak, or non-edges. Strong edges are kept, weak edges are kept if they are connected to strong edges, and non-edges are discarded.

### hysteresis procedure: 
1. it is the final step in the edge detection algorithm.
2. it is used to decide which edge is real and which are noise using two threshold values.
3. The Canny edge detection algorithm uses a technique called hysteresis to connect edges. It works by first applying two thresholds to the gradient magnitude of the image.
4. strong edges : pixels with value greater than upper threshold --> keep them as strong edge and is included in the final edge map.
5. weak edge : If a pixel's gradient magnitude is between the lower and upper thresholds, it is considered a weak edge and is only included in the final edge map if it is connected to a strong edge. and  if not connected to the strong edge then remove that.
6. below the lower threshold: remove completely (considered as noise)
7. This helps to reduce noise and improve the accuracy of edge detection.

### sobel operator:
1. the sobel operator is an edge detection method used to find the gradient  (rate of intensity change) in an image
2. it helps to detect : 1. vertical edges    2. horizontal edges  ==> by applying small filters to the image
3. edge occurs where there is sudden change in pixel intensity
4. sobel calculates:
5. 1. change in the x- direction
    2. change in the y- direction

```python
image = cv2.imread("D:\\VS Code\\Computer_Vision\\cameraman.png")

grey_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Canny edge detection
# The parameters 100 and 200 are the lower and upper thresholds for the hysteresis procedure.
edged_image = cv2.Canny(grey_image, 100, 200)

cv2.imshow("Original Image", image)
cv2.imshow("Grey Image", grey_image)
cv2.imshow("Edged Image", edged_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Activity 02: Image Blurring using openCV

```python
image = cv2.imread("D:\\VS Code\\Computer_Vision\\cameraman.png")

guassian_blurred_image = cv2.GaussianBlur(image, (15, 15), 0)

median_blurred_image = cv2.medianBlur(image, 15)
# median blur removes the salt and pepper noise, while Gaussian blur is more effective for reducing Gaussian noise. The choice between the two depends on the specific characteristics of the noise in the image and the desired outcome.

cv2.imshow("Original Image", image)
cv2.imshow("Gaussian Blurred Image", guassian_blurred_image)   
cv2.imshow("Median Blurred Image", median_blurred_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
