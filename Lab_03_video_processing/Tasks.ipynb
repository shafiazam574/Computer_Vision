# Task 01:


```python
import cv2

# ------------------------------
# Lab Task 1: Processing Video Frames
# ------------------------------

# Load the video
video = cv2.VideoCapture("D:\\VS Code\\Computer_Vision\\Lab_03_video_processing\\video_01.mp4")

if not video.isOpened():
    print("Error: Could not open video.")
    exit()

# Get video properties
width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(video.get(cv2.CAP_PROP_FPS))

frame_count = 0

while True:
    ret, frame = video.read()
    if not ret:
        break  # End of video

    frame_count += 1

    # ------------------------------
    # Image Processing Techniques
    # ------------------------------

    # Resize frame (50%)
    resized_frame = cv2.resize(frame, (width//2, height//2))

    #  Apply Gaussian Blur
    blurred_frame = cv2.GaussianBlur(frame, (15, 15), 0)

    # 3️⃣ Convert to Grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_frame_bgr = cv2.cvtColor(gray_frame, cv2.COLOR_GRAY2BGR)  # convert back to BGR for concatenation

    # ------------------------------
    # Resize all images to same size for 2x2 display
    # ------------------------------
    half_width = width // 2
    half_height = height // 2

    frame_small = cv2.resize(frame, (half_width, half_height))
    resized_small = cv2.resize(resized_frame, (half_width, half_height))
    blurred_small = cv2.resize(blurred_frame, (half_width, half_height))
    gray_small = cv2.resize(gray_frame_bgr, (half_width, half_height))

    # ------------------------------
    # Combine images into 2x2 layout
    # ------------------------------
    top_row = cv2.hconcat([frame_small, resized_small])
    bottom_row = cv2.hconcat([blurred_small, gray_small])
    combined_display = cv2.vconcat([top_row, bottom_row])

    # ------------------------------
    # Display the combined frames
    # ------------------------------
    cv2.imshow("Lab Task 1: Original & Processed Frames", combined_display)

    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
print(f"Lab Task 1 Completed: {frame_count} frames processed and displayed successfully.")

```

# Task 02:

```python
import cv2

# ------------------------------
# Lab Task 2: Write Processed Frames to Video
# ------------------------------

#  Load video
video = cv2.VideoCapture("D:\\VS Code\\Computer_Vision\\Lab_03_video_processing\\video_01.mp4")

if not video.isOpened():
    print("Error: Could not open video.")
    exit()

#  Get video properties
width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(video.get(cv2.CAP_PROP_FPS))

#  Create VideoWriter object
out_video = cv2.VideoWriter(
    "processed_video_lab2.mp4",
    cv2.VideoWriter_fourcc(*'mp4v'),  # Codec
    fps,                              # Frames per second
    (width, height)                   # Frame size
)

frame_count = 0

while True:
    ret, frame = video.read()
    if not ret:
        break  # End of video

    frame_count += 1

    # ------------------------------
    # Image Processing Techniques
    # ------------------------------

    # Resize frame (50%)
    resized_frame = cv2.resize(frame, (width//2, height//2))

    # Apply Gaussian Blur
    blurred_frame = cv2.GaussianBlur(frame, (15, 15), 0)

    # Convert to Grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_frame_bgr = cv2.cvtColor(gray_frame, cv2.COLOR_GRAY2BGR)

    # ------------------------------
    # Resize all images to same size for 2x2 layout
    # ------------------------------
    half_width = width // 2
    half_height = height // 2

    frame_small = cv2.resize(frame, (half_width, half_height))
    resized_small = cv2.resize(resized_frame, (half_width, half_height))
    blurred_small = cv2.resize(blurred_frame, (half_width, half_height))
    gray_small = cv2.resize(gray_frame_bgr, (half_width, half_height))

    # ------------------------------
    # Combine images into 2x2 layout
    # ------------------------------
    top_row = cv2.hconcat([frame_small, resized_small])
    bottom_row = cv2.hconcat([blurred_small, gray_small])
    combined_frame = cv2.vconcat([top_row, bottom_row])

    # ------------------------------
    # Write combined frame to video
    # ------------------------------
    out_video.write(combined_frame)

    # Optional: Display progress
    if frame_count % 30 == 0:
        print(f"Processed {frame_count} frames...")

video.release()
out_video.release()
cv2.destroyAllWindows()
print(f"Lab Task 2 Completed: Video saved as 'processed_video_lab2.mp4' ({frame_count} frames).")

```

```python
import cv2

# ------------------------------
# Lab Task 3: Add Dynamic Text on Video
# ------------------------------

# 1️⃣ Load video
video = cv2.VideoCapture("D:\\VS Code\\Computer_Vision\\Lab_03_video_processing\\video_01.mp4")

if not video.isOpened():
    print("Error: Could not open video.")
    exit()

# 2️⃣ Get video properties
width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(video.get(cv2.CAP_PROP_FPS))

# 3️⃣ Create VideoWriter object
out_video = cv2.VideoWriter(
    "video_with_text_lab3.mp4",
    cv2.VideoWriter_fourcc(*'mp4v'),  # Codec
    fps,                              # Frames per second
    (width, height)                   # Frame size
)

frame_count = 0

while True:
    ret, frame = video.read()
    if not ret:
        break  # End of video

    frame_count += 1

    # ------------------------------
    # Image Processing Techniques
    # ------------------------------

    # Resize frame (50%)
    resized_frame = cv2.resize(frame, (width//2, height//2))

    # Apply Gaussian Blur
    blurred_frame = cv2.GaussianBlur(frame, (15, 15), 0)

    # Convert to Grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_frame_bgr = cv2.cvtColor(gray_frame, cv2.COLOR_GRAY2BGR)

    # ------------------------------
    # Resize all images to same size for 2x2 layout
    # ------------------------------
    half_width = width // 2
    half_height = height // 2

    frame_small = cv2.resize(frame, (half_width, half_height))
    resized_small = cv2.resize(resized_frame, (half_width, half_height))
    blurred_small = cv2.resize(blurred_frame, (half_width, half_height))
    gray_small = cv2.resize(gray_frame_bgr, (half_width, half_height))

    # ------------------------------
    # Combine images into 2x2 layout
    # ------------------------------
    top_row = cv2.hconcat([frame_small, resized_small])
    bottom_row = cv2.hconcat([blurred_small, gray_small])
    combined_frame = cv2.vconcat([top_row, bottom_row])

    # ------------------------------
    # Add dynamic text
    # ------------------------------
    text = f"Frame: {frame_count}"
    cv2.putText(
        combined_frame,            # Frame to write text on
        text,                      # Text string
        (50, 50),                  # Position (x, y)
        cv2.FONT_HERSHEY_SIMPLEX,  # Font
        1,                         # Font scale
        (0, 0, 255),               # Red color (BGR)
        2                          # Thickness
    )

    # ------------------------------
    # Write processed frame with text to video
    # ------------------------------
    out_video.write(combined_frame)

    # Optional: Show progress every 30 frames
    if frame_count % 30 == 0:
        print(f"Processed {frame_count} frames...")

video.release()
out_video.release()
cv2.destroyAllWindows()
print(f"Lab Task 3 Completed: Video with text saved as 'video_with_text_lab3.mp4' ({frame_count} frames).")
```
