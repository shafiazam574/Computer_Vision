# Activity 01:
    Play a Video using OpenCV

```python
# video = cv2.VideoCapture("E:\\vs code\\git\\Computer Vision\\Lab_03_video_processing\\video_01.mp4")
video = cv2.VideoCapture("D:\\VS Code\\Computer_Vision\\Lab_03_video_processing\\Recreated_video_01.mp4")

# Check if the video was opened successfully
if not video.isOpened():
    print("Error opening video file")

# Read and display video frames
while True:
    # video.read() returns a boolean (ret) and the frame itself (frame)
    ret, frame = video.read()
    if not ret:
        print("End of video stream")
        break
    cv2.imshow("Video Frame", frame)

    # Wait for 25 ms and check if the user has pressed the 'q' key to exit
    # ord() converts the character 'q' to its ASCII value, which is used for comparison
    if cv2.waitKey(25) & 0xFF == ord('q'):
        print("Video playback interrupted by user")
        break
    
# Release the video capture object and close all OpenCV windows that is opened during the function cv2.VideoCapture() and cv2.imshow()
video.release()
cv2.destroyAllWindows()
```

# extracting images from the video_01

```python
video = cv2.VideoCapture("D:\\VS Code\\Computer_Vision\\Lab_03_video_processing\\video_01.mp4")

# Check if the folder exists, if not create it
if not os.path.exists("frames for A_02"):
    os.mkdir("frames for A_02")

count = 0

while True:
    ret, frame = video.read()
    if not ret:
        print("End of video stream")
        break
    cv2.imwrite(f"frames for A_02/frame_{count}.jpg", frame)
    count += 1

video.release()
print(f"Extracted {count} frames from the video and saved to 'frames for A_02' directory.")
cv2.destroyAllWindows()
```

# Activity 02:
    Create Video using Multiple Images

```python
image_folder = "frames for A_02"

# Sort images based on the frame number extracted from the filename
# and then store the sorted list of image filenames in the variable 'images'
images = sorted(os.listdir(image_folder), key=lambda x: int(x.split('_')[1].split('.')[0]))

# read the first image to get dimensions/size of the video frames
# join() is used to concatenate the folder path and the image filename to get the full path of the image
frame = cv2.imread(os.path.join(image_folder, images[0]))
height, width, layers = frame.shape


# Create a VideoWriter object to write the video file
# cv2.videoWriter() takes the output filename, codec, frame rate, and frame size as arguments
# codec is specified using cv2.VideoWriter_fourcc() function, where 'mp4v' is the codec for MP4 format. and we use * for unpacking the string 'mp4v' into individual characters as required by the function.
video = cv2.VideoWriter(
    "Recreated_video_01.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),  # Codec for MP4 format
    30,  # Frame rate (adjust as needed)
    (width, height)  # Size of each frame
)

for image in images:
    img = cv2.imread(os.path.join(image_folder, image))
    video.write(img)

video.release()
print("Video recreated successfully as 'Recreated_video_01.mp4'.") 
```
